## 2. The question: decision, action, cost of a wrong call

**Decision:** Which signals should we prioritize for monitoring or feature engineering?  
**Action:** Engineers and analysts use this to decide which metrics to track, which to include in ranking models, and where to invest instrumentation effort.  
**Cost of a wrong call:** Ignoring a strong signal wastes potential lift; including noise signals wastes compute and obscures true patterns. This is signal analysis, not a causal claim — we're measuring correlation and variance, not intervention.

## 3. Quick look at the data (2-3 real numbers)

The signals show real variance: position ranges 0–245, days with impressions 1–88, CTR 0–100. About 54% of pages are declining. This spread suggests signals have discriminatory power — some pages rank high on certain metrics while declining, others do not. A signal audit can quantify which signals separate declining from stable pages most strongly.

```python
from pathlib import Path
import pandas as pd

repo_root = Path.cwd().parent.parent
data_path = repo_root / "data" / "raw" / "content_refresh_anonymized.csv"
df = pd.read_csv(data_path)

df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)

print(f"Rows: {len(df):,}")
print(f"Declining-label rate: {df['is_declining_label'].mean():.3f}")
print(f"Average days with impressions: {df['days_with_impressions'].mean():.2f}")
print(f"Average avg_position: {df['avg_position'].mean():.2f}")
```

## 4. Careful words: what I can and can't claim

**I can claim:**  
- Observed: Which signals correlate with declining pages, and by how much (effect sizes, variance, means).
- Measured: Grouped comparisons (declining vs. stable pages) with honest base rates and distributions.
- Directional: Signal A is a stronger predictor than signal B on this dataset.

**I cannot claim:**  
- Causal: Changing a signal causes decline (or prevents it).
- Predictive beyond holdout: Generalization to future unseen data without proper time-aware validation.
- Universal: "This signal always matters" without context on content type, client, or time period.

```python
from pathlib import Path
import pandas as pd

repo_root = Path.cwd().parent.parent
data_path = repo_root / "data" / "raw" / "content_refresh_anonymized.csv"
df = pd.read_csv(data_path)

df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)
base_rate = df["is_declining_label"].mean()

print(f"Observed base rate: {base_rate:.3f}")
print("This supports a decision-support ranking claim, not a causal claim.")
```
